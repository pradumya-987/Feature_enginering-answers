## 1. What is a parameter?

## 5. How does loss value help in determining whether the model is good or not?
Loss measures how far the predicted output is from the actual output. Lower loss usually means better learning on the given data. Comparing training and testing loss also helps identify overfitting or underfitting.

## 6. What are continuous and categorical variables?
- Continuous variables: Numeric values that can take any value in a range, such as height or salary.
- Categorical variables: Variables that represent groups or labels, such as city, gender, or product type.

## 7. How do we handle categorical variables in Machine Learning? What are the common techniques?
Categorical variables are converted into numerical form before model training.

Common techniques:
- Label Encoding
- One-Hot Encoding
- Ordinal Encoding
- Frequency Encoding
- Target Encoding

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import pandas as pd

sample = pd.DataFrame({"Color": ["Red", "Blue", "Green", "Red"]})

label_encoder = LabelEncoder()
sample["Color_Label"] = label_encoder.fit_transform(sample["Color"])

one_hot = pd.get_dummies(sample["Color"], prefix="Color")

sample, one_hot
```

## 11. How do we split data for model fitting (training and testing) in Python?

```python
from sklearn.model_selection import train_test_split
import pandas as pd

X = pd.DataFrame({"x1": [1,2,3,4,5], "x2": [10,20,30,40,50]})
y = pd.Series([0,1,0,1,0])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

X_train, X_test, y_train, y_test
```

## 17. What is causation? Explain difference between correlation and causation with an example.
Causation means one event directly produces another event. Correlation only shows association, not direct cause.

Example: Ice cream sales and cold drink sales may both rise in summer. They are correlated, but one does not cause the other. Hot weather affects both.

## 18. What is an Optimizer? What are different types of optimizers? Explain each with an example.
An optimizer updates model parameters to reduce the loss function.

Common types:
- Gradient Descent: Uses the whole dataset for each update.
- Stochastic Gradient Descent (SGD): Uses one record at a time.
- Mini-Batch Gradient Descent: Uses a small batch of records.
- Momentum: Uses previous gradients to speed convergence.
- RMSProp: Adapts learning rates using recent squared gradients.
- Adam: Combines Momentum and RMSProp.

## 19. What is sklearn.linear_model?
`sklearn.linear_model` is a Scikit-learn module that contains linear models such as LinearRegression, LogisticRegression, Ridge, Lasso, and ElasticNet.

## 20. What does model.fit() do? What arguments must be given?

```python
#Only 2 arguements are required i.e. xtrain and ytrain below is the example

from sklearn.linear_model import LinearRegression
import pandas as pd

X_train = pd.DataFrame({"Experience": [1, 2, 3, 4, 5]})
y_train = pd.Series([25000, 30000, 35000, 40000, 45000])

model = LinearRegression()
model.fit(X_train, y_train)

model
```

## 21. What does model.predict() do? What arguments must be given?

```python
X_test = pd.DataFrame({"Experience": [6, 7]})
predictions = model.predict(X_test)
predictions
```

model.predict(X_test) uses the trained model to generate predictions for new input data.

## 23. What is feature scaling? How does it help in Machine Learning?
Feature scaling means transforming numeric features to a common scale. It helps algorithms converge faster and prevents features with large numeric ranges from dominating the model.

## 24. How do we perform scaling in Python?

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import pandas as pd

X = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Income": [20000, 30000, 50000, 70000, 90000]
})

standard_scaler = StandardScaler()
X_standard_scaled = pd.DataFrame(standard_scaler.fit_transform(X), columns=X.columns)

minmax_scaler = MinMaxScaler()
X_minmax_scaled = pd.DataFrame(minmax_scaler.fit_transform(X), columns=X.columns)

X_standard_scaled, X_minmax_scaled
```
